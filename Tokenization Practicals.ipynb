```python
corpus = """Hello there you are in python environment. 
This is github repository NLP and DL learning. Happy Learning! """
```

```python
corpus
```

```python
## Tokenization
##paragraph --> sentences
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize
```

#### punkt_tab 
It is a tokenizer data package for the Natural Language Toolkit (NLTK) 
in Python. It is a newer, tab-based version of the classic punkt tokenizer data,
designed to improve performance and accuracy when splitting text 
into sentences and words.

```python
documents =sent_tokenize(corpus)
```

```python
for sentence in documents:
    print(sentence)
```

```python
type(documents)
```

```python
##Tokenization
##Paragraph -->Words
##Sentences -->Words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
# This will also split punctuations like 'User's' 
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
#In this type of Tokenizer FUll STOP(.) is not separated from word
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordDetokenizer
```

```python
tknz = TreebankWordDetokenizer()
#This splits the word with whitespaces and also catches line breaks
```

```python
tknz.tokenize(corpus)
```
