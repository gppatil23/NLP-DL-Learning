```python
sentence="The Eiffel Tower was built from 1887 to 1889 by French engineerGustave Eiffel, whose company specialized in building metal frameworks and structures."

'''
Person Eg: Gaurav Patil
Place Or Location Eg: India
Date Eg: September, 24-09-1989
Time Eg: 4:30pm
Money Eg: 1 million dollar
Organization Eg: iNeuron Private Limited
Percent Eg: 20%, twenty percent
    
'''
```

### With NLP we can recognize precise type(entity) of words
1. Person Eg: Gaurav Patil
2. Place Or Location Eg: India
3. Date Eg: September, 24-09-1989
4. Time Eg: 4:30pm
5. Money Eg: 1 million dollar
6. Organization Eg: iNeuron Private Limited
7. Percent Eg: 20%, twenty percent

```python
import nltk
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
words = nltk.word_tokenize(sentence)
```

```python
tag_element = nltk.pos_tag(words)
```

```python
nltk.ne_chunk(tag_element).draw()
```
